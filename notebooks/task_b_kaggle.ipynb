# Task B: Project MANAS reproduction launcher

Select a CUDA GPU and Internet access. Add authorized data as private Kaggle inputs. This launcher uses release tag `v0.1.0` and defaults to planning one group. Set the input paths and `EXECUTE=True` only when ready to train. CPU source tests do not establish GPU compatibility. Keep tokens in Kaggle Secrets, never in notebook source.


```python
from pathlib import Path
import os, subprocess, sys
REPO_URL = "https://github.com/robinpnalex/ProjectMANAS-HastikaICON2026.git"
RELEASE_REF = "v0.1.0"
WORK = Path("/kaggle/working/ProjectMANAS-HastikaICON2026")
if not WORK.exists():
    subprocess.run(["git", "clone", "--depth", "1", "--branch", RELEASE_REF, REPO_URL, str(WORK)], check=True)
expected = subprocess.check_output(["git", "-C", str(WORK), "rev-parse", RELEASE_REF], text=True).strip()
actual = subprocess.check_output(["git", "-C", str(WORK), "rev-parse", "HEAD"], text=True).strip()
if actual != expected:
    raise RuntimeError("Existing checkout differs from the requested release")
os.chdir(WORK)
subprocess.run([sys.executable, "-m", "pip", "install", "-e", str(WORK) + "[llm]"], check=True)
```

```python
RAW_DIR = Path("/kaggle/input/SET-YOUR-PRIVATE-DATASET/raw")
EXTERNAL = Path("/kaggle/input/SET-YOUR-PRIVATE-DATASET/offenseval_kn.csv")
GROUP = "run38_session1"
EXECUTE = False
command = [sys.executable, "-m", "hastika.runner", "--config", "configs/task_b/submitted.json",
           "--raw-dir", str(RAW_DIR), "--external", str(EXTERNAL), "--groups", GROUP]
if EXECUTE:
    command.append("--execute")
subprocess.run(command, check=True)
```
