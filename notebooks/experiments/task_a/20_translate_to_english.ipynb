## 1. Spelling-robust slur canonicaliser

**Phonetic skeleton.** Every token is reduced to a key that is the same across the common
ways of romanizing one Kannada word: leetspeak undone (`s00le`), obfuscation stripped
(`s*le`), emphatic lengthening collapsed (`suuuule`), long vowels and aspirates merged
(`soole`, `thika`, `bhosadi`), doubled consonants merged (`sulle`, `keyya`), `w`/`v` and
`z`/`j` merged.

**Matching.** Exact skeleton match; then a slur prefix followed by a known Kannada suffix
(`soolemagane`); then edit distance 1, only for keys of five letters or more, because
short keys collide with real words. Adjacent tokens are also tried joined, so `boli maga`
matches `bolimaga`. English dictionary words never match, which keeps `sure`, `sale` and
`tune` out.

**The seed lexicon is a starting point for review.** It is restricted to words that are
abusive on their own. Words that are only abusive in context, such as `maga` (son), `gandu`
(male) or caste names, are left to the translator. The diagnostic cell prints every surface
spelling each entry caught and how often those rows are labelled Hate. Prune with `BLOCK`
anything that looks wrong.

````python
# IndicXlit depends on fairseq -> omegaconf<2.1, whose legacy metadata is rejected by pip >=24.1.
# Kaggle's current Python image therefore needs the pip compatibility fix before install.
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "pip==24.0"], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "--no-cache-dir",
                "ai4bharat-transliteration==1.1.0.1", "IndicTransToolkit", "nltk"],
               check=True)
import re
from collections import Counter, defaultdict

import nltk
from rapidfuzz.distance import Levenshtein

nltk.download("words", quiet=True)
from nltk.corpus import words as nltk_words

EN_VOCAB = {w.lower() for w in nltk_words.words()}

# English gloss -> romanized spellings. Spellings only need to cover distinct skeletons;
# the skeleton already absorbs vowel length, aspiration, doubling and leetspeak.
LEXICON = {
    "whore": ["sule", "randi"],
    "son of a whore": ["sulemaga", "sulemagane", "randimaga"],
    "bitch": ["munde"],
    "son of a bitch": ["bolimaga", "bolimagane"],
    "bastard": ["bevarsi"],
    "cunt": ["bosudi", "bosadi", "bosdi"],
    "fuck": ["keya"],
    "dick": ["tunne", "lavda", "lowde", "lowda"],
    "ass": ["thika"],
    "dog": ["nayi"],
    "eunuch": ["chakka"],
    "filthy": ["kachada", "kachda"],
}
# Skeletons that must never match: everyday words a fuzzy match reached in the first
# Kaggle run. "kannada" was glossed "filthy", "gowda" "dick", "gandhi" "whore" and
# "chikka" (small) "eunuch".
NEVER = ["kannada", "kanada", "kanda", "khanda", "gowda", "gouda", "howda", "howdu",
         "gandhi", "gandi", "gandu", "gaandu", "chikka", "rochaka", "turka", "kunne"]
BLOCK = set()        # further skeletons to never match, after reviewing the diagnostics
SUFFIXES = ["maga", "magane", "magne", "mage", "makalu", "maklu", "makla", "galu", "ge"]

LEET = str.maketrans("013457@$!", "oieastasi")
DIGRAPHS = [("aa", "a"), ("ee", "e"), ("ii", "i"), ("oo", "u"), ("uu", "u"), ("ou", "u"),
            ("ph", "f"), ("bh", "b"), ("ch", "c"), ("dh", "d"), ("gh", "g"), ("kh", "k"),
            ("sh", "s"), ("th", "t"), ("ck", "k"), ("w", "v"), ("z", "j"), ("q", "k")]
THU = re.compile(r"^t+h+[uo]+$", re.I)   # "thu"/"thooo": spitting, "shame on you"
EDGE = re.compile(r"^([^\w@$*]*)(.*?)([^\w*]*)$")


def skeleton(tok):
    t = tok.lower().translate(LEET)
    t = re.sub(r"[^a-z]", "", t)
    t = re.sub(r"(.)\1{2,}", r"\1\1", t)           # suuuule -> suule
    t = re.sub(r"([^aeiou])\1+", r"\1", t)         # chhakka -> chaka, sulle -> sule
    for a, b in DIGRAPHS:
        t = t.replace(a, b)
    return re.sub(r"(.)\1+", r"\1", t)


KEYS = {}
for gloss, spellings in LEXICON.items():
    for s in spellings:
        KEYS[skeleton(s)] = gloss
SEED_KEYS = set(KEYS)                    # fuzzy matching is allowed against these only
BLOCK |= {skeleton(w) for w in NEVER}
SUFFIX_KEYS = {skeleton(s) for s in SUFFIXES}


def match(core):
    """(gloss, matched key, leftover suffix) for one token, or None."""
    if THU.match(core):
        return "shame on you", "tu", ""
    if core.lower() in EN_VOCAB:
        return None
    sk = skeleton(core)
    if len(sk) < 3 or sk in BLOCK:
        return None
    if sk in KEYS:
        return KEYS[sk], sk, ""
    for k in KEYS:
        if len(k) >= 4 and sk.startswith(k) and sk[len(k):] in SUFFIX_KEYS:
            return KEYS[k], k, sk[len(k):]
    if len(sk) >= 6:
        for k in SEED_KEYS:
            if len(k) >= 6 and k[:3] == sk[:3] and Levenshtein.distance(sk, k, score_cutoff=1) <= 1:
                return KEYS[k], k, ""
    return None


def canonicalize(text, mode="gloss"):
    """mode="gloss": slurs become English glosses (before translation).
    mode="canon": slurs become one canonical romanized spelling (for char models).
    Returns (text, [(surface, key), ...])."""
    toks = text.split()
    parts = [EDGE.match(t).groups() for t in toks]
    out, hits, i = [], [], 0
    while i < len(toks):
        pre, core, post = parts[i]
        if i + 1 < len(toks) and core and parts[i + 1][1]:
            nxt = parts[i + 1][1]
            if core.lower() not in EN_VOCAB and nxt.lower() not in EN_VOCAB:
                sk = skeleton(core + nxt)
                if sk in KEYS and sk not in BLOCK:
                    out.append(pre + (KEYS[sk] if mode == "gloss" else sk) + parts[i + 1][2])
                    hits.append((core + " " + nxt, sk))
                    i += 2
                    continue
        m = match(core) if core else None
        if m is None:
            out.append(toks[i])
        else:
            gloss, key, rest = m
            word = gloss if mode == "gloss" else key
            out.append(pre + word + (" " + rest if rest else "") + post)
            hits.append((core, key))
        i += 1
    return " ".join(out), hits


for probe in ["nin sule maga", "SOOOLE magane", "s00le", "bewarsi nan maga", "boli maga thoo",
              "sure bro", "sulekha is here"]:
    print(f"  {probe!r:24} -> {canonicalize(probe)[0]!r}")
````

### Semantic expansion with fastText

Spelling rules miss variants they were not written for. A fastText model with character
n-gram subwords is trained on the **unlabelled** comment text (the training rows only, labels unused;
holdout text is never read). For each lexicon spelling, its nearest neighbours are candidate
variants. A candidate is accepted only when both tests pass: cosine similarity of at least
0.80, and a skeleton that is either one edit from the seed with the same two opening letters, or the seed plus a suffix. Semantic closeness alone pulls in
co-occurring words such as `maga` next to `sule`, and spelling closeness alone pulls in
unrelated words. Accepted additions are printed for review.

````python
TOKEN = re.compile(r"[a-z0-9@$*]+")
sentences = [TOKEN.findall(t.lower()) for t in tr["clean"]]   # training rows only
added = {}
try:
    from gensim.models import FastText
    ft = FastText(sentences, vector_size=100, window=4, min_count=2, min_n=2, max_n=5,
                  sg=1, epochs=30, seed=0, workers=1)
    for gloss, spellings in LEXICON.items():
        for s in spellings:
            if s not in ft.wv.key_to_index:
                continue
            for w, sim in ft.wv.most_similar(s, topn=40):
                sk = skeleton(w)
                seed = skeleton(s)
                # a respelling (one edit, same opening) or the seed plus a suffix --
                # never just "close in meaning", which is how gowda, kunne and gandi got in
                close = ((Levenshtein.distance(sk, seed) <= 1 and sk[:2] == seed[:2])
                         or (len(seed) >= 4 and sk.startswith(seed)))
                if (sim >= 0.80 and len(sk) >= 4 and sk not in KEYS and sk not in BLOCK
                        and w not in EN_VOCAB and close):
                    added[sk] = (gloss, w, round(float(sim), 3))
    for sk, (gloss, w, sim) in sorted(added.items(), key=lambda kv: kv[1][0]):
        KEYS[sk] = gloss
        print(f"  + {w:16} ({sk:12}) -> {gloss:16} sim {sim}")
    print(f"fastText added {len(added)} variant skeletons; lexicon now {len(KEYS)}")
except ImportError as e:
    print("gensim unavailable, spelling rules only:", e)
````

### Diagnostics: what the lexicon catches

These counts describe the data. They do not select entries, since choosing entries by label
rate would leak labels into the folds. A gloss whose matched rows are mostly Non-Hate is a
candidate for `BLOCK`.

````python
y_all = (tr["Label"] == "Hate").values
tr["glossed"], tr_hits = zip(*[canonicalize(t) for t in tr["clean"]])
ho["glossed"] = [canonicalize(t)[0] for t in ho["clean"]]
tr["has_slur"] = [bool(h) for h in tr_hits]

surfaces, rows_by_gloss = defaultdict(Counter), defaultdict(set)
for r, hits in enumerate(tr_hits):
    for surface, key in hits:
        g = "shame on you" if key == "tu" else KEYS[key]
        surfaces[g][surface.lower()] += 1
        rows_by_gloss[g].add(r)
print(f"rows with a lexicon hit: {tr['has_slur'].mean():.1%} "
      f"(Hate rate in them {y_all[tr['has_slur']].mean():.1%}, overall {y_all.mean():.1%})\n")
for g in sorted(rows_by_gloss, key=lambda g: -len(rows_by_gloss[g])):
    rows = list(rows_by_gloss[g])
    print(f"{g:16} rows {len(rows):5}  hate {y_all[rows].mean():.0%}  "
          f"spellings {len(surfaces[g]):3}: {', '.join(w for w, _ in surfaces[g].most_common(12))}")
````

## 2. Translation backends

Both translators sit behind one interface, `translate(list_of_str) -> list_of_str`. Either may
fail to load; a failure is reported and that backend is skipped.

| backend | pipeline | needs a token? |
|---|---|---|
| `sarvamai/sarvam-translate` | Gemma-3-4B fine-tuned for Indic -> English, code-mixed text included; reads the comment directly | no |
| IndicXlit -> `ai4bharat/indictrans2-indic-en-1B` | transliterates to Kannada script, then the strongest dedicated Kannada -> English model | yes: `HF_TOKEN`, license accepted once |

`bodhan-ai/indic-translate` is dropped. It is gated, and it needs transformers 5.x, which
IndicTrans2 cannot run on.

Sarvam-Translate runs in bf16 because Gemma-3 overflows in fp16. On a T4, bf16 is
emulated: slower, but correct. Garbage output would show in the gate's empty-output rate.

````python
DTYPE = torch.bfloat16 if torch.cuda.get_device_capability(0)[0] >= 8 else torch.float16


class SarvamTranslate:
    """sarvamai/sarvam-translate: Gemma-3-4B fine-tuned by Sarvam AI with AI4Bharat for the
    22 scheduled languages, code-mixed input included. Ungated, so no token is needed."""
    name = "sarvam-translate"
    MODEL = "sarvamai/sarvam-translate"
    PROMPT = "Translate the text below to English."

    def __init__(self):
        from transformers import AutoModelForCausalLM, AutoTokenizer
        self.tok = AutoTokenizer.from_pretrained(self.MODEL)
        self.tok.padding_side = "left"
        # Gemma-3 overflows in fp16; bf16 is emulated on a T4, slower but correct
        dtype = torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16
        self.model = AutoModelForCausalLM.from_pretrained(
            self.MODEL, torch_dtype=dtype, device_map={"": 0}).eval()

    def __call__(self, texts, bs=16):
        out = [""] * len(texts)
        order = np.argsort([-len(t) for t in texts])   # similar lengths share a batch
        for i in range(0, len(order), bs):
            idx = order[i:i + bs]
            prompts = [self.tok.apply_chat_template(
                [{"role": "system", "content": self.PROMPT},
                 {"role": "user", "content": texts[j][:700]}],
                add_generation_prompt=True, tokenize=False) for j in idx]
            enc = self.tok(prompts, return_tensors="pt", padding=True,
                           add_special_tokens=False).to(self.model.device)
            with torch.no_grad():
                gen = self.model.generate(**enc, max_new_tokens=192, do_sample=False)
            dec = self.tok.batch_decode(gen[:, enc["input_ids"].shape[1]:],
                                        skip_special_tokens=True)
            for j, d in zip(idx, dec):
                out[j] = d.strip()
        return out

    def close(self):
        del self.model
        gc.collect()
        torch.cuda.empty_cache()


class IndicTrans2:
    """IndicXlit (romanized -> Kannada script) then IndicTrans2 1B (Kannada -> English).
    Word-level language ID masks handles, numerals, emoji and English words, including the
    injected glosses, so they pass through both models unchanged."""
    name = "indictrans2-1B"
    MODEL = "ai4bharat/indictrans2-indic-en-1B"
    NEUTRAL = re.compile(r"^(?:@\w+|#\w+|<\w+>|https?://\S+|[\d.,:%/+-]+|[^\w\s]+)$")

    def __init__(self):
        # Pin what is already loaded: a fairseq-era dependency that wants to downgrade
        # numpy, torch or transformers must fail here, not break the running session.
        import importlib.metadata as md
        pathlib.Path("/tmp/pins.txt").write_text("".join(
            f"{p}=={md.version(p)}\n" for p in ("numpy", "torch", "transformers")))
        subprocess.run("pip install -q -c /tmp/pins.txt ai4bharat-transliteration "
                       "IndicTransToolkit", shell=True, check=True)
        from ai4bharat.transliteration import XlitEngine
        from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
        try:
            from IndicTransToolkit import IndicProcessor
        except ImportError:
            from IndicTransToolkit.processor import IndicProcessor
        self.xlit = XlitEngine("kn", beam_width=4, rescore=False)
        self.tok = AutoTokenizer.from_pretrained(self.MODEL, trust_remote_code=True)
        self.model = AutoModelForSeq2SeqLM.from_pretrained(
            self.MODEL, trust_remote_code=True, torch_dtype=torch.float16).to("cuda").eval()
        self.ip = IndicProcessor(inference=True)
        probe = self.to_kannada("nin sule maga")
        assert re.search(r"[ಀ-೿]", probe), f"IndicXlit gave no Kannada: {probe!r}"

    def to_kannada(self, text):
        kept, pieces = {}, []
        for i, tok in enumerate(text.split()):
            if (self.NEUTRAL.match(tok) or re.search(r"[ಀ-೿]", tok)
                    or tok.lower() in EN_VOCAB):
                kept[f"@@{i}@@"] = tok
                pieces.append(f"@@{i}@@")
            else:
                pieces.append(tok)
        out = self.xlit.translit_sentence(" ".join(pieces), lang_code="kn") if pieces else ""
        if isinstance(out, dict):
            out = out.get("kn") or next(iter(out.values()))
        if isinstance(out, (list, tuple)):
            out = out[0]
        out = str(out)
        for ph, tok in kept.items():
            out = out.replace(ph, tok)
        return out

    def __call__(self, texts, bs=32):
        kn = [self.to_kannada(t) for t in texts]
        done = []
        for i in range(0, len(kn), bs):
            chunk = self.ip.preprocess_batch(kn[i:i + bs], src_lang="kan_Knda",
                                             tgt_lang="eng_Latn")
            enc = self.tok(chunk, truncation=True, padding=True, max_length=256,
                           return_tensors="pt").to("cuda")
            with torch.no_grad():
                gen = self.model.generate(**enc, num_beams=5, max_length=256)
            done += self.ip.postprocess_batch(
                self.tok.batch_decode(gen, skip_special_tokens=True), lang="eng_Latn")
        return done

    def close(self):
        del self.model
        gc.collect()
        torch.cuda.empty_cache()


BACKENDS = [SarvamTranslate, IndicTrans2]
````

## 3. Stage 1: which translator keeps the abuse, and does the signal survive?

The same 300 lexicon-matched comments go through each backend twice:

- **raw**: the comment as written. This measures the translator alone: does it render
  abuse as abuse, or launder it?
- **protected**: slurs replaced by English glosses first. This is what the pipeline will
  actually use.

The backend with the higher raw survival wins, because raw survival also reflects
everything the lexicon does not cover. The gate is then applied to the winner's protected
output: if the glosses themselves do not survive, or many outputs come back empty, the run
stops.

````python
EN_ABUSE = re.compile(
    r"\b(fuck\w*|shit\w*|bitch\w*|bastard\w*|whore\w*|slut\w*|prostitut\w*|cunt\w*|dick\w*|"
    r"ass|asshole\w*|eunuch\w*|idiot\w*|stupid|nonsense|rascal\w*|scoundrel\w*|dogs?|"
    r"bloody|damn\w*|disgust\w*|shame\w*|filth\w*|useless|nasty|abus\w*|son of a)\b", re.I)

sample = tr[tr["has_slur"]].sample(min(300, int(tr["has_slur"].sum())), random_state=0)
gate = {}
for Backend in BACKENDS:
    try:
        b = Backend()
    except Exception as e:
        print(f"[{Backend.name}] could not load, skipped: {type(e).__name__}: {e}")
        continue
    try:
        raw = b(sample["clean"].tolist())
        prot = b(sample["glossed"].tolist())
    except Exception as e:
        print(f"[{Backend.name}] failed while translating, skipped: {type(e).__name__}: {e}")
        continue
    finally:
        b.close()
    gate[Backend.name] = {
        "raw_survival": float(np.mean([bool(EN_ABUSE.search(e)) for e in raw])),
        "protected_survival": float(np.mean([bool(EN_ABUSE.search(e)) for e in prot])),
        "empty": float(np.mean([len(e.strip()) < 5 for e in prot])),
    }
    print(f"\n[{Backend.name}] {gate[Backend.name]}")
    for i in range(8):
        print("  romanized :", sample["clean"].iloc[i][:100])
        print("  raw       :", raw[i][:100])
        print("  protected :", prot[i][:100])
        print()

assert gate, "no translation backend could be loaded"
BEST = max(gate, key=lambda k: (gate[k]["raw_survival"] - gate[k]["empty"]))
print("chosen backend:", BEST, gate[BEST])
````

````python
GATE = 0.50     # on protected output; the glosses are put there deliberately, so it should be high
g = gate[BEST]
if g["raw_survival"] < 0.25:
    print(f"warning: raw survival {g['raw_survival']:.1%} -- the translator launders abuse on "
          "its own, so the pipeline depends on the lexicon for slurs it does not list")
if g["protected_survival"] < GATE or g["empty"] > 0.2:
    raise SystemExit("GATE FAILED: abuse does not survive even with glossary protection, or "
                     "outputs are empty. Record it and stop.")
print("gate passed, continuing")
````

## 4. Stage 2: translate all 7,193 rows

Cached under `artifacts/translations/`, keyed by backend, lexicon hash and holdout
fingerprint, so editing `LEXICON` or `BLOCK` re-translates rather than reusing stale text.

Two **separate copies** of the repo are prepared, and the git clone is never modified:

| copy | holds | role |
|---|---|---|
| `hastika_en` | translated training rows; translated holdout as the inference file | hateBERT |
| `hastika_rom` | the same rows romanized | MuRIL, the baseline |

The trainer always predicts its inference file, so a full fit on the training rows writes
holdout probabilities with no source change.

````python
Backend = next(B for B in BACKENDS if B.name == BEST)
lex_hash = hashlib.md5(json.dumps([sorted(KEYS.items()), sorted(BLOCK), HOLDOUT_FP,
                                   getattr(Backend, "PROMPT", "")]).encode()).hexdigest()[:8]
CACHE = pathlib.Path("artifacts/translations"); CACHE.mkdir(parents=True, exist_ok=True)
paths = {name: CACHE / f"{BEST}_{lex_hash}_{name}.csv" for name in ("train", "holdout")}

if not all(p.exists() for p in paths.values()):
    b = Backend()
    for name, df in [("train", tr), ("holdout", ho)]:
        if paths[name].exists():
            continue
        out = df[["id", "Label"]].copy()
        out.insert(1, "Comment", b(df["glossed"].tolist()))
        out.to_csv(paths[name], index=False)
        print("wrote", paths[name], len(out))
    b.close()
else:
    print("reusing", *paths.values())

eng_tr = pd.read_csv(paths["train"]).fillna({"Comment": ""})
eng_ho = pd.read_csv(paths["holdout"]).fillna({"Comment": ""})
assert (eng_tr["id"].values == tr["id"].values).all() and (eng_ho["id"].values == ho["id"].values).all()


def workdir(path, train_df, infer_df):
    shutil.rmtree(path, ignore_errors=True)
    shutil.copytree(WORK, path, ignore=shutil.ignore_patterns(".git", "artifacts"))
    train_df[["id", "Comment", "Label"]].to_csv(f"{path}/data/raw/binary_train.csv", index=False)
    infer_df[["id", "Comment"]].to_csv(f"{path}/data/raw/binary_validation_inputs.csv", index=False)
    return path


EN_WORK = workdir("/kaggle/working/hastika_en", eng_tr, eng_ho)
ROM_WORK = workdir("/kaggle/working/hastika_rom", tr, ho)
print(eng_tr.shape, eng_ho.shape)
for a, b_ in zip(tr["clean"].head(5), eng_tr["Comment"].head(5)):
    print(" ", a[:80], "\n   ->", b_[:80])
````

## 5. Stage 3: hateBERT on English and MuRIL on romanized, side by side

Each arm trains on the 6,114 training rows (`--folds 1`) with 3 seeds averaged and predicts
the 1,079 holdout rows. The two run in parallel, one per T4.

- **hateBERT**, `GroNLP/hateBERT`, on the translated English text.
- **MuRIL**, Run 9's component recipe on the romanized text: demojized, two re-initialised
  layers, 6 epochs.

`--no-dedupe` is required. The rows are already deduplicated, and deduplicating again on
the *English* text would merge distinct comments that happen to translate identically.

````python
subprocess.run('pip install -q "transformers>=4.45,<5"', shell=True, check=True)
from hastika.common.gpu_queue import run_queue

SEEDS = ["42", "43", "44"]
COMMON = ["--folds", "1", "--seeds", *SEEDS, "--epochs", "6", "--bs", "16", "--eval-bs", "64",
          "--reinit-layers", "2", "--no-dedupe"]
ARMS = {"a_en_hatebert": (EN_WORK, ["--model", "GroNLP/hateBERT", "--no-demojize"]),
        "a_rom_muril": (ROM_WORK, ["--model", "google/muril-base-cased"])}
jobs = [{"name": tag, "cwd": work, "env": {"PYTHONPATH": f"{work}/src"},
         "log": f"{WORK}/artifacts/logs/{tag}.log",
         "cmd": " ".join([sys.executable, "-u", "-m", "hastika.models.muril", "--tag", tag,
                          *extra, *COMMON])}
        for tag, (work, extra) in ARMS.items()]
codes = run_queue(jobs, raise_on_fail=False)
print("exit codes:", codes)
````

## 6. Stage 4: the holdout comparison

Every arm is scored on the same 1,079 rows, with fixed weights only.

**Paired bootstrap:** both arms are scored on the same resampled rows 5,000 times, which
removes the row-sampling noise they share. The questions it answers:

- Does hateBERT on English beat MuRIL on romanized, as a single model?
- Does adding hateBERT improve the current-style SVM + MuRIL blend?
- Does adding hateBERT improve the SVM alone?

The char SVM is also scored on **canonicalised** romanized text: every slur variant is
rewritten to one spelling. This tests the spelling normaliser on its own, independently
of translation.

````python
from sklearn.calibration import CalibratedClassifierCV
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, f1_score
from sklearn.pipeline import make_pipeline
from sklearn.svm import LinearSVC

y_tr = (tr["Label"] == "Hate").astype(int).values
y = (ho["Label"] == "Hate").astype(int).values


def holdout_probs(work, tag):
    p = pathlib.Path(work) / "artifacts/runs" / tag / "test_probs.npy"
    if not p.exists():
        print(f"{tag}: no predictions, see {WORK}/artifacts/logs/{tag}.log")
        return None
    out = np.load(p)
    assert out.shape == (len(y), 2), out.shape
    return out


def svm_probs(X_train, X_hold):
    m = make_pipeline(TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=2,
                                      sublinear_tf=True),
                      CalibratedClassifierCV(LinearSVC(C=0.5, class_weight="balanced"), cv=3))
    return m.fit(X_train, y_tr).predict_proba(X_hold)


Xtr = np.array([clean(t, demojize=True) for t in tr["Comment"]])
Xho = np.array([clean(t, demojize=True) for t in ho["Comment"]])
canon = lambda X: np.array([canonicalize(t, mode="canon")[0] for t in X])
svm = svm_probs(Xtr, Xho)
svm_canon = svm_probs(canon(Xtr), canon(Xho))
en = holdout_probs(EN_WORK, "a_en_hatebert")
mu = holdout_probs(ROM_WORK, "a_rom_muril")

arms = {"char SVM, romanized": svm, "char SVM, canonicalised": svm_canon}
if mu is not None:
    arms["MuRIL, romanized (3 seeds)"] = mu
    arms["SVM 0.57 + MuRIL 0.43"] = 0.57 * svm + 0.43 * mu
if en is not None:
    arms[f"hateBERT, English via {BEST} (3 seeds)"] = en
    arms["SVM + hateBERT, 50/50"] = (svm + en) / 2
    if mu is not None:
        arms["SVM + MuRIL + hateBERT, thirds"] = (svm + mu + en) / 3

scores = {}
print(f"{'arm':44s} {'macro-F1':>8s} {'acc':>6s} {'Hate F1':>8s}")
for k, p in arms.items():
    pred = p.argmax(1)
    scores[k] = float(f1_score(y, pred, average="macro"))
    print(f"{k:44s} {scores[k]:8.4f} {accuracy_score(y, pred):6.4f} {f1_score(y, pred):8.4f}")
print("references, older measurements on other rows: TF-IDF floor 0.8073 OOF, TAPT MuRIL "
      "0.8128 OOF, Run 9 blend 0.8213 combined OOF")

boots = {}
pairs = []
if en is not None and mu is not None:
    pairs += [("hateBERT - MuRIL", en, mu),
              ("thirds - (SVM+MuRIL)", arms["SVM + MuRIL + hateBERT, thirds"],
               arms["SVM 0.57 + MuRIL 0.43"])]
if en is not None:
    pairs += [("(SVM+hateBERT) - SVM", arms["SVM + hateBERT, 50/50"], svm)]
pairs += [("canonicalised SVM - SVM", svm_canon, svm)]
print()
for label, a, b in pairs:
    r = paired_bootstrap(y, a, b)
    boots[label] = r
    print(f"{label:26s} {r['diff']:+.4f}  95% CI [{r['ci95'][0]:+.4f}, {r['ci95'][1]:+.4f}]  "
          f"P(better) {r['p_better']:.2f}")

dis = either = None
if en is not None:
    dis = float((en.argmax(1) != svm.argmax(1)).mean())
    either = float(((en.argmax(1) == y) | (svm.argmax(1) == y)).mean())
    print(f"\nhateBERT and SVM disagree on {dis:.1%} of holdout rows; either is right on {either:.1%}")
````

## 7. Preserve outputs

The translated CSVs are the expensive part and the reusable one. Keep them even if the model
loses: any later idea on translated text starts from them for free. The final lexicon,
including fastText additions, is saved so it can be reviewed and committed.

````python
OUT = pathlib.Path("/kaggle/working/task_a_english_outputs")
OUT.mkdir(parents=True, exist_ok=True)
for p in paths.values():
    shutil.copy2(p, OUT / p.name)
for f in pathlib.Path("artifacts/logs").glob("*.log"):
    shutil.copy2(f, OUT / f.name)
for name, p in [("svm", svm), ("svm_canon", svm_canon), ("muril", mu), ("hatebert_en", en)]:
    if p is not None:
        np.save(OUT / f"holdout_probs_{name}.npy", p)
ho[["id", "Label"]].to_csv(OUT / "holdout_ids.csv", index=False)
json.dump({"keys": KEYS, "block": sorted(BLOCK),
           "fasttext_added": {k: list(v) for k, v in added.items()}},
          open(OUT / "slur_lexicon.json", "w"), indent=2, ensure_ascii=False)
json.dump({"commit": head, "holdout_fingerprint": HOLDOUT_FP, "backend": BEST, "gate": gate,
           "exit_codes": codes, "holdout_macro_f1": scores, "paired_bootstrap": boots,
           "disagreement": dis, "either_right": either},
          open(OUT / "result.json", "w"), indent=2)
print(sorted(x.name for x in OUT.iterdir()))
````

## 8. What to do with the result

Record in `docs/EXPERIMENTS.md`, including a gate failure:
- the chosen backend and both survival rates
- the holdout table
- the paired-bootstrap lines

How to read the outcomes:

- **Gate fails.** Translation launders the abuse even with glossary protection. Record it
  and stop.
- **`thirds - (SVM+MuRIL)` has an interval above 0.** Translation earns a place as a third
  ensemble member. That is the result that would justify a full-data submission with it.
- **`hateBERT - MuRIL` above 0.** A genuinely better single representation, attacking the
  tokenizer problem at its root.
- **`canonicalised SVM - SVM` above 0.** The spelling normaliser is worth adding to the
  main TF-IDF pipeline on its own, whatever happens to translation.
